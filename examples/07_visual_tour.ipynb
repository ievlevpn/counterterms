# A visual tour of `counterterms`

Run the cells top to bottom. Every object below draws itself.

```python
from sympy import Derivative
from counterterms import SPDE, Noise, Parabolic, Unknown, playground, build_renormalization

# KPZ:  (∂t − Δ) u = (∂x u)² + ξ,  spacetime white noise in d = 1
u  = Unknown("u", dim=1)
xi = Noise.white("xi", dim=1)
kpz = SPDE(operator=Parabolic(dim=1), unknown=u, noises=[xi],
           rhs=xi.symbol + Derivative(u.field, u.x[0])**2)
eq = kpz.renormalize()
eq
```

## One tree

One tree of the family, then its extraction coproduct `δτ` and twisted antipode `S'₋τ`.

```python
rn = build_renormalization(kpz)
t = eq.all_trees[2]
eq.show(t)
```

```python
rn.coaction(t)
```

```python
rn.twisted_antipode(t)
```

```python
print(rn.coaction(t))        # the same sum as text
```

## Hand-built trees

The playground needs no SPDE.

```python
pg = playground("-3/2 - kappa")
cherry = pg.tree(pg.I(pg.Xi, "x"), pg.I(pg.Xi, "x"))
cherry
```

```python
pg.delta_minus(cherry)
```

## Why a coefficient is 2

`cuts` lists the cuts of a tree behind each coproduct term. Extracting one noise from the cherry has coefficient 2: two mirror-image cuts, drawn here with the extracted node shaded.

```python
from counterterms.render import svg
from counterterms.render.display import Picture
from counterterms.trees.coproducts import cuts

key = next(k for k in cuts(cherry, pg.sig) if k[0] == (pg.Xi,))
Picture(" ".join(svg(cherry, pg.sig, scale=1.6, number=True,
                     mark={i: g for g, block in enumerate(blocks) for i in block})
                 for _coeff, blocks in cuts(cherry, pg.sig)[key]))
```
